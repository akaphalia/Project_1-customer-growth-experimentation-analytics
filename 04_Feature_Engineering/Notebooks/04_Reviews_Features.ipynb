# order_reviews Feature Engineering

## Objective

This notebook creates business-oriented features from the cleaned `order_reviews` dataset.

```python
import pandas as pd
import numpy as np

order_reviews = pd.read_csv(
    "../../03_Python_ETL/Output/order_reviews_clean.csv",
    parse_dates=[
        "review_creation_date",
        "review_answer_timestamp"
    ]
)
```

### Feature 1: review_response_time_days

**Business Purpose:**
Calculates the time taken to respond to a customer review. This helps evaluate customer support responsiveness and monitor review handling performance.

```python
order_reviews["review_response_time_days"] = (
    order_reviews["review_answer_timestamp"]
    - order_reviews["review_creation_date"]
).dt.total_seconds() / (24 * 3600)
```

```python
# Validation
order_reviews[
    [
        "review_creation_date",
        "review_answer_timestamp",
        "review_response_time_days"
    ]
].head()
```

```python
order_reviews["review_response_time_days"].describe()
```

```python
order_reviews["review_response_time_days"].isnull().sum()
```

```python
(order_reviews["review_response_time_days"] < 0).sum()
```

### Feature 2: has_review_comment

**Business Purpose:**
Identifies whether a customer wrote a review comment along with the review score. This helps distinguish between rating-only reviews and reviews containing written customer feedback, which are valuable for sentiment and qualitative analysis.

```python
order_reviews["has_review_comment"] = (
    order_reviews["review_comment_message"]
    .notna()
    .astype(int)
)
```

```python
# Validation
order_reviews[
    [
        "review_comment_message",
        "has_review_comment"
    ]
].head(10)
```

```python
order_reviews["has_review_comment"].value_counts()
```

```python
order_reviews["has_review_comment"].isnull().sum()
```

### Feature 3: positive_review_flag

**Business Purpose:**
Identifies whether a review is positive based on the review score. This helps measure customer satisfaction and enables quick analysis of positive customer experiences across products, sellers, and time periods.

```python
order_reviews["positive_review_flag"] = (
    order_reviews["review_score"] >= 4
).astype(int)
```

```python
# Validation
order_reviews[
    [
        "review_score",
        "positive_review_flag"
    ]
].head(10)
```

```python
order_reviews["positive_review_flag"].value_counts()
```

```python
order_reviews["positive_review_flag"].isnull().sum()
```
