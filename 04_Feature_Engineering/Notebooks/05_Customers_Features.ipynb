# customers Feature Engineering

## Objective

This notebook creates business-oriented features from the cleaned `customers` dataset.

```python
import pandas as pd
import numpy as np

customers = pd.read_csv(
    "../../03_Python_ETL/Output/customers_clean.csv"
)
```

```python
customers.head()
```

### Feature 1: customer_city_state
**Business Purpose**

Creates a combined geographic identifier (city + state) that is much more useful for reporting than separate columns. It simplifies grouping, filtering, and visualization in SQL and Power BI.

```python
customers["customer_city_state"] = (
    customers["customer_city"] + ", " +
    customers["customer_state"]
)
```

```python
# Validation
customers[
    [
        "customer_city",
        "customer_state",
        "customer_city_state"
    ]
].head(10)
```
