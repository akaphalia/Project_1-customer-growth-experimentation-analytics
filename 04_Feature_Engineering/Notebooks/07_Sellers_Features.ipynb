# sellers Feature Engineering

## Objective

This notebook creates business-oriented features from the cleaned `sellers` dataset.

```python
import pandas as pd

sellers = pd.read_csv(
    "../../03_Python_ETL/Output/sellers_clean.csv"
)
```

### Feature 1: seller_city_state
**Business Purpose**

Creates a combined geographic identifier for each seller.

This makes grouping, filtering, reporting, and mapping much easier in SQL and Power BI than working with city and state separately.

```python
sellers["seller_city_state"] = (
    sellers["seller_city"] + ", " +
    sellers["seller_state"]
)
```

```python
#Validation
sellers[
    [
        "seller_city",
        "seller_state",
        "seller_city_state"
    ]
].head()
```
