# products Feature Engineering

## Objective

This notebook creates business-oriented features from the cleaned `products` dataset.

```python
import pandas as pd
import numpy as np

products = pd.read_csv(
    "../../03_Python_ETL/Output/products_clean.csv"
)
```

### Feature 1: product_volume_cm3
**Business Purpose**

Calculates the physical volume of each product in cubic centimeters. This is useful for logistics, warehouse storage, shipping analysis, and understanding how product size affects freight costs and delivery performance.

```python
products["product_volume_cm3"] = (
    products["product_length_cm"] *
    products["product_height_cm"] *
    products["product_width_cm"]
)
```

```python
# Validation
products[
    [
        "product_length_cm",
        "product_height_cm",
        "product_width_cm",
        "product_volume_cm3"
    ]
].isnull
```
